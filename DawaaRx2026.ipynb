```python
import pandas as pd
month_names = df['month_name'].unique()
for month in month_names:
    msk = df['month_name'] == month
    df_region_status = df[msk].groupby(['region', 'status'])['status'].size().reset_index(name='count')
    
    total_per_region = (
        df_region_status
        .groupby('region')['count']
        .sum()
        .reset_index(name='Total Orders')
    )
    
    # Order Collected
    collected = (
        df_region_status[df_region_status['status'] == 'Order Collected']
        .groupby('region')['count']
        .sum()
        .reset_index(name='Collected Orders')
    )
    
    # Order Cancelled (لو عندك أكتر من نوع إلغاء)
    cancelled = (
        df_region_status[
            df_region_status['status'].str.contains('Cancelled|Reject|INVALID', case=False)
        ]
        .groupby('region')['count']
        .sum()
        .reset_index(name='Cancelled Orders')
    )
    pending = (
        df_region_status[
            ~df_region_status['status'].str.contains('Cancelled|Reject|INVALID|Collected', case=False)
        ]
        .groupby('region')['count']
        .sum()
        .reset_index(name='Pending Orders'))
    net_value = df[msk].groupby('region')['netvalue'].sum('netvalue').reset_index().sort_values(by="netvalue" , ascending = False)
    # دمج كل النتائج
    summary = (
        total_per_region
        .merge(collected, on='region', how='left')
        .merge(cancelled, on='region', how='left')
        .merge(pending, on='region', how='left')
        .merge(net_value, on='region', how='left')
        .fillna(0)
    )
    
    # حساب النسب
    summary['order_collected_%'] = (
        summary['Collected Orders'] / summary['Total Orders'] * 100
    ).round(2)
    
    summary['order_cancelled_%'] = (
        summary['Cancelled Orders'] / summary['Total Orders'] * 100
    ).round(2)
    
    summary['order_pending_%'] = (
        summary['Pending Orders'] / summary['Total Orders'] * 100
    ).round(2)
    
    summary = summary.sort_values(
        by='Total Orders',
        ascending=False
    ).reset_index(drop=True)
    
    summary_with_total = summary.copy()
    
    grand_total = pd.DataFrame([{
        'region': 'GRAND TOTAL',
        'Total Orders': summary['Total Orders'].sum(),
        'Collected Orders': summary['Collected Orders'].sum(),
        'order_collected_%': (
            summary['Collected Orders'].sum() /
            summary['Total Orders'].sum() * 100
        ).round(2),
        'Cancelled Orders': summary['Cancelled Orders'].sum(),
        'order_cancelled_%': (
            summary['Cancelled Orders'].sum() /
            summary['Total Orders'].sum() * 100
        ).round(2),
        'Pending Orders': summary['Pending Orders'].sum(),
        'order_pending_%': (
            summary['Pending Orders'].sum() /
            summary['Total Orders'].sum() * 100
        ).round(2),
        'netvalue': summary['netvalue'].sum(),
    }])
    
    summary_with_total = pd.concat(
        [summary_with_total, grand_total],
        ignore_index=True
    )
    summary_with_total.to_excel(f'rx_region_{month}.xlsx')
```
